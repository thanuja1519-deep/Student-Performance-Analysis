```python
df = pd.read_csv('StudentsPerformance.csv')
df.head()
df.head(10)
df.shape
df.columns
df.info()
df.isnull().sum()
df.duplicated().sum()
df = df.drop_duplicates()
df[['math score', 'reading score', 'writing score']].mean()
df[['math score', 'reading score', 'writing score']].max()
df[['math score', 'reading score', 'writing score']].min()
df[['math score', 'reading score', 'writing score']].describe()
df.groupby('test preparation course')[
    ['math score', 'reading score', 'writing score']
].mean()
df.groupby('parental level of education')[
    ['math score', 'reading score', 'writing score']
].mean()
df['reading score'].corr(df['writing score'])
sns.scatterplot(
    data=df,
    x='reading score',
    y='writing score'
)

plt.title('Reading vs Writing Scores')
plt.show()
```

**Question**: Which subject has lowest average score?

```python
average_scores = df[
    ['math score', 'reading score', 'writing score']
].mean()
plt.figure(figsize=(8, 5))

average_scores.plot(kind='bar')
plt.title('Average Scores by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Marks')
plt.xticks(rotation=0)
plt.show()
```

By observing the bar graph, we can see that the **math score** has lower average score


**Question**: How are the math scores of students distributed

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='math score',
    bins=10,
    kde=True
)

plt.title('Distribution of Mathematics Scores')
plt.xlabel('Mathematics Marks')
plt.ylabel('Number of Students')

plt.show()
```

The distribution curve displays that more than 250 students have scored around the mean and the least score is much scattered from mean with very few students.

**Question**:How strongly are Mathematics, Reading and Writing scores related to each other?

```python
correlation = df[
    ['math score', 'reading score', 'writing score']
].corr()
plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Between Subject Scores')

plt.show()
```

Using this correlation heatmap, we can understand that reading and writing scores are highly correlated to each other and writing and math are relatively less correlated to each other
